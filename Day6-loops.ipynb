## Loops

```python
# for loop we can use to iterate over sequence of items, list, tupel, range dictionary
a= "python"
for i in a:
    print(i)
    
    
```

```python
# for with list
number = ["aws", "azure", "gcp"]
for i in number:
    print(i)
```

```python
# for 
number = [1,2,3,4,5,6]
for i in number:
       print(i)
```

```python
# for with break
number = [1,2,3,4,5,6]
for i in number:
   if i == 3:
       break
   print(i)
```

```python
# for with dictionary
number = {
    "name" : "santosh",
    "course" : "FDE"
}
for i in number:
    print(i)
```

```python
# for with dictionary items ( name and values)
number = {
    "name" : "santosh",
    "course" : "FDE"
}
for i in number.items():
    print(i)
```

```python
# for with dictionary values
number = {
    "name" : "santosh",
    "course" : "FDE"
}
for i in number.values():
    print(i)
```

```python
# for with dictionary with keys
number = {
    "name" : "santosh",
    "course" : "FDE"
}
for i in number.keys():
    print(i)
```

```python

total =1
dev = range(1,5)
for i in dev:
    total = total+i
print("total=", total)
```

```python
# for with if else
number = [1,2,3,4,5,6]
for i in number:
   if i%2 == 0:
       print(i)
   else:
       print("odd number")
```

```python
# for with if else,7
number = [10,-5, 0, 20, -8]
for num in number:
   if num > 0:
       print(num, "is positive")
   elif num < 0:
       print(num, "is negative")
   else:
       print(num, "is zero")
```
