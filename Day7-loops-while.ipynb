## while loop
### while loop is used to repeat a block of code as log as given condition is true
### while loop can use index (i) to access list of elements
### for loop is used to iterate over the sequence of elements ( list, tuple, dictionery)

```python
#infinity loop
while True:
    print("hello")
```

```python
i=1
while True:
    print(i)
```

```python
i=1
while i <= 10:
    print(i)
    i=i+1   #increment operator
```

```python
i=1
while i in range (1,4):
    print(i)
    i=i+1
```

```python
i=1
while i in range (1,4):
    password= input ("Enter your password: ")
    if password == "san":
        print("allow to login")
    else:
        print("not allow")
    i=i+1
```

```python
i=1
while i in range (1,5):
    password= input ("Enter your password: ")
    if password == "san":
        print("allow to login")
        break
    else:
        print("not allow")
    i=i+1
```

```python
## for loop example
for i in range (1,4):
    question = input("ASK to AI: ")
    if(question.lower() == "exit"):
        print("chatbot stopped")
        break
    print("AI is responding")
```

```python
## while loop with exit and break example
while True:
    question = input("ASK to AI: ")
    if(question.lower() == "exit"):
        print("chatbot stopped")
        break
    print("AI is responding")
```

```python
## for loop example with continue and break
for i in range (1,6):   
    if i == 2:
        continue   # it will skip further execution and continue next loop
    if i == 4:
        break  # it will skip complete loop
    print(i)
```
