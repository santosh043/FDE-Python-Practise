## Python Conditions if, elif, else, Nested condition

### Rule-1: use colon (:) every if, else, elif
### Rule-2: Indentation mandatory
### Rule-3: elif, else cannot stand alone
### correct indentitation example
     if age>18:
        print("adult")

```python
#simple if
age = 20
if age >= 18:
    print("you are elegible to vote")
```

```python
#simple if and else
age = 17
if age >= 18:
    print("you are elegible to vote")
else:
    print("You are not eligible to vote")
```
